# craggy-charts in Jupyter

Each chart is a widget, so it works in JupyterLab, Notebook, VS Code and Colab.

```bash
pip install craggy-charts pandas
```

````python
import math
import pandas as pd
import craggy_charts as cc

days = pd.date_range("2026-01-01", periods=120, freq="D", name="date")
df = pd.DataFrame({
    "web": [round(1200 + 400 * math.sin(i / 14) + 6 * i) for i in range(120)],
    "app": [round(700 + 250 * math.sin(i / 9 + 1) + 4 * i) for i in range(120)],
}, index=days)

visits = cc.chart(df, type="line", title="Daily visits", zoom="x")
visits
````

## Morph

`update()` takes any options, including a new type, and the tiles animate to it.

````python
visits.update(type="area", stack=True, cell="chunky")
````

## Events

Click a point, drag to zoom (double-click resets) or toggle the legend. Events arrive in Python.

````python
from ipywidgets import Output

log = Output()
visits.on("click", lambda p: log.append_stdout(f"click {p['title']}\n"))
visits.on("zoom", lambda e: log.append_stdout(f"zoom {e['from']} → {e['to']}\n"))
visits.on("legend", lambda e: log.append_stdout(f"legend {e['name']} hidden={e['hidden']}\n"))
log
````

## Live data

`window=60` keeps the last 60 rows; `push()` appends and the chart scrolls.

````python
import random, time

cpu = cc.chart(type="area", window=60, y=["api", "worker"], format="percent", y_axis={"min": 0, "max": 100}, title="CPU")
display(cpu)

for t in range(120):
    cpu.push({"t": str(t), "api": 45 + 25 * math.sin(t / 7) + random.random() * 10,
              "worker": 30 + 15 * math.sin(t / 11) + random.random() * 8})
    time.sleep(0.25)
````

## Export

`save()` writes one self-contained HTML file that opens with no network.

````python
visits.save("visits.html")
````
